````python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Redéfinir num_columns à partir de X_train pour exclure 'fraude'
num_columns = X_train.select_dtypes(include=[np.number]).columns.tolist()

# Création du pipeline numérique autonome
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Redéfinir les colonnes catégorielles en excluant les identifiants et 'resultat_enquete'
cat_columns = [col for col in X_train.select_dtypes(exclude=[np.number]).columns.tolist() if col not in ['id_operation', 'id_client', 'resultat_enquete']]

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])
preprocessor = ColumnTransformer(transformers=[
    ('num', num_pipeline, num_columns),
    ('cat', cat_pipeline, cat_columns)
])


X_train_prepared = preprocessor.fit_transform(X_train)
X_val_prepared = preprocessor.transform(X_test)
````

###  la différence :

* **StandardScaler** utilise la **moyenne** et l'**écart-type** :

  Comme la moyenne et l'écart-type sont fortement influencés par les valeurs extrêmes (outliers), ces dernières tirent la moyenne vers le haut et augmentent artificiellement l'écart-type. En conséquence, les valeurs ordinaires se retrouvent écrasées et très regroupées près de zéro.

* **RobustScaler** utilise la **médiane** et l'**écart interquartile (IQR)** :
  La médiane et l'IQR ne sont pas influencés par les valeurs aberrantes. Ainsi, l'échelle reste représentative de la majorité des données normales :
  * Une transaction ordinaire (égale à la médiane) aura toujours une valeur exacte de **0**.
  * Les valeurs extrêmes conservent des scores très élevés sans perturber ou écraser l'échelle des transactions ordinaires.

###  résultats : Q1, Q3 et IQR

* **Q1 (Premier Quartile = 384.27 DH) :**
  Cela signifie que **25 %** des transactions ont un montant inférieur ou égal à 384.27 DH. Les 75 % restants ont un montant supérieur.

* **Q3 (Troisième Quartile = 1244.90 DH) :**
  Cela signifie que **75 %** des transactions ont un montant inférieur ou égal à 1244.90 DH. Seules 25 % des transactions dépassent ce montant.

* **IQR (Écart Interquartile = 860.63 DH) :**
  L'IQR mesure la dispersion de la **moitié centrale** de vos données (entre le 25e et le 75e percentile). C'est la différence entre Q3 et Q1 ($1244.90 - 384.27 = 860.63$).
  * **Ce qu'il mesure :** Contrairement à l'écart-type, l'IQR est une mesure de dispersion **robuste**. Il nous montre la plage dans laquelle se situent 50 % des transactions ordinaires (le cœur des données), sans être perturbé ou faussé par les transactions extrêmement élevées (outliers).


### Ces montants extrêmes sont-ils nécessairement des erreurs ?

Non, les montants situés au-dessus de $L2 = 2535.85$ DH ne sont pas nécessairement des erreurs de saisie ou des anomalies techniques. Voici pourquoi :

1. **Nature des transactions :** Dans le domaine bancaire ou commercial, il est tout à fait normal d'avoir une distribution asymétrique vers la droite. La majorité des achats quotidiens sont de faibles montants (inférieurs à 1200 DH), mais des achats importants et parfaitement légitimes (électroménager, voyages, abonnements annuels) peuvent facilement dépasser 2500 DH.
2. **Lien avec la fraude :** Certains montants très élevés peuvent être le signe de transactions frauduleuses plutôt que de simples erreurs de saisie.
3. **Seuil statistique vs Réalité métier :** La règle de $1.5 \times IQR$ est une méthode purement statistique et générique pour identifier des valeurs atypiques dans une distribution normale, mais elle ne prend pas en compte le contexte métier de l'application.


````python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Stratégie 1 : Conserver les montants d'origine (X_train['montant_dh'])
# 2. Stratégie 2 : Limiter (clipper) les montants aux seuils L1 et L2
# Comme L1 est négatif (-906.68) et qu'un montant ne peut pas être négatif, le seuil bas logique est 0.
montant_clipped = X_train['montant_dh'].clip(lower=0, upper=L2)

# Création de la figure pour comparer les deux stratégies
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Boxplot d'origine
sns.boxplot(ax=axes[0], x=X_train['montant_dh'], color='skyblue')
axes[0].set_title("Option A : Conservation des montants extrêmes (Original)")
axes[0].set_xlabel("Montant (DH)")

# Boxplot après limitation (clipping)
sns.boxplot(ax=axes[1], x=montant_clipped, color='lightgreen')
axes[1].set_title(f"Option B : Limitation au seuil L2 ({L2:.2f} DH)")
axes[1].set_xlabel("Montant (DH) - Limité")

plt.tight_layout()
plt.show()
````

### Pourquoi faut-il justifier ce choix de traitement ?

Le choix entre conserver ou limiter (clipper) les valeurs extrêmes a un impact direct sur la qualité de votre modèle et doit être justifié selon les arguments suivants :

#### 1. Impact sur les modèles sensibles aux échelles (StandardScaler)
* Si on utilise **StandardScaler** et qu'on **conserve** les valeurs extrêmes (Option A), la moyenne et l'écart-type seront artificiellement tirés vers le haut. Cela va écraser toutes les transactions ordinaires et réduire la capacité du modèle à distinguer les nuances dans 94 % des données.
* Si on **limite** les montants (Option B), on stabilise la variance, ce qui rend l'algorithme plus performant et robuste.

#### 2. Pertinence métier (Détection de la fraude)
* **Conserver (Option A)** est souvent indispensable dans la détection de fraude. En effet, les transactions frauduleuses se cachent très fréquemment derrière des montants inhabituels ou extrêmement élevés. Supprimer ou altérer cette information (Option B) revient à masquer le signal même de la fraude.
* **Limiter (Option B)** est utile si l'on suspecte que les valeurs extrêmes ne sont que des erreurs de saisie (bruit) sans valeur informationnelle.

#### 3. Choix de l'algorithme
* Si vous utilisez des modèles basés sur les arbres de décision (comme Random Forest ou XGBoost), ces derniers ne sont pas sensibles à l'échelle ni aux outliers. Conserver les valeurs d'origine (Option A) est alors souvent le meilleur choix car cela évite de perdre de l'information sans nuire au modèle.

### 28 & 29. Création et application du ColumnTransformer final

````python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler, OneHotEncoder
import numpy as np

# 1. Pipeline numérique final (Imputation par la médiane + RobustScaler)
final_num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', RobustScaler())
])

# 2. Pipeline catégoriel final (Imputation par valeur fréquente + OneHotEncoder)
final_cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

# 3. Assembleur final
preprocessor_final = ColumnTransformer(transformers=[
    ('num', final_num_pipeline, num_columns),
    ('cat', final_cat_pipeline, cat_columns)
])

# Ajustement uniquement sur le groupe de référence (sans fuite)
X_train_final = preprocessor_final.fit_transform(X_train)
X_test_final = preprocessor_final.transform(X_test)

# Vérifications
print(f"Dimensions de X_train_final : {X_train_final.shape}")
print(f"Dimensions de X_test_final  : {X_test_final.shape}")
print(f"Valeurs manquantes restantes dans X_train_final : {np.isnan(X_train_final).sum()}")
print(f"Valeurs manquantes restantes dans X_test_final  : {np.isnan(X_test_final).sum()}")
````

#### Rôle de `handle_unknown='ignore'` :
Lorsqu'un modèle rencontre une catégorie inconnue en phase de test ou de production, l'encodage One-Hot classique lève une erreur car il n'a pas de colonne correspondante pré-allouée. En spécifiant `handle_unknown='ignore'`, le transformateur ignore simplement cette valeur inconnue et lui attribue des zéros (`0`) sur l'ensemble des colonnes associées à cette variable, évitant ainsi le plantage de l'application.

````python
import joblib

# Enregistrement du transformateur
model_path = 'preprocessor_final.joblib'
joblib.dump(preprocessor_final, model_path)
print(f"Transformateur enregistré sous : {model_path}")

# Rechargement
loaded_preprocessor = joblib.load(model_path)

# Test de cohérence sur un petit échantillon
echantillon = X_test.iloc[:5]
res_original = preprocessor_final.transform(echantillon)
res_charge = loaded_preprocessor.transform(echantillon)

assert np.allclose(res_original, res_charge), "Erreur : Les résultats diffèrent !"
print("Vérification réussie : Le transformateur rechargé produit exactement les mêmes résultats sans recalcul.")
````
